$$\delta(x) = \frac{P\,x^2\,(3L - x)}{6\,E\,I}$$

```python
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Datos (unidades del SI)

```python
L = 3.0  # longitud [m]
P = 10e3  # carga en el extremo [N]
E = 210e9  # módulo de elasticidad del acero [Pa]
I = 1943e-8  # inercia de un IPE 200 [m⁴]
```

## 2. Cálculo de la flecha

```python
x = np.linspace(0, L, 100)
flecha = P * x**2 * (3 * L - x) / (6 * E * I)
print(f"Flecha máxima: {flecha.max() * 1000:.1f} mm")
```

## 3. Gráfico

Cambia `P` o `L` en la celda de datos, vuelve a ejecutar y mira cómo cambia la curva.

```python
plt.plot(x, -flecha * 1000)
plt.xlabel("x [m]")
plt.ylabel("flecha [mm]")
plt.title("Viga en voladizo: IPE 200, P = 10 kN")
plt.grid(True)
plt.show()
```
