# Función de dos variables en 3D

Dibuja $f(x, y) = \sin x \cos y$ sobre $[0, a] \times [0, b]$ y calcula su integral doble
con la regla de Simpson. Mueve los deslizadores `a` y `b` para cambiar el dominio.

Librerías: `conda install -n profesia numpy scipy plotly ipywidgets`

```python
import ipywidgets as widgets
import numpy as np
import plotly.graph_objects as go
from IPython.display import display
from scipy.integrate import simpson


# Función de dos variables
def f(x, y):
    """Función que se dibuja e integra."""
    return np.sin(x) * np.cos(y)


def actualizar(a, b):
    """Dibuja f en [0, a] × [0, b] y pone su integral doble en el título."""
    x = np.linspace(0, a, 100)
    y = np.linspace(0, b, 100)
    X, Y = np.meshgrid(x, y)
    Z = f(X, Y)

    # Integral doble en [0,a] × [0,b]
    integral_y = simpson(Z, x=y, axis=0)
    integral = simpson(integral_y, x=x)

    fig = go.Figure(data=[go.Surface(x=X, y=Y, z=Z, colorscale="Viridis")])
    fig.update_layout(
        title=f"f(x, y) = sin(x) cos(y) | Integral ≈ {integral:.6f}",
        scene=dict(xaxis_title="x", yaxis_title="y", zaxis_title="f(x, y)"),
        margin=dict(l=0, r=0, t=50, b=0),
    )
    display(fig)


slider_a = widgets.FloatSlider(value=3.0, min=0.1, max=10.0, step=0.1, description="a:")
slider_b = widgets.FloatSlider(value=3.0, min=0.1, max=10.0, step=0.1, description="b:")

panel = widgets.interactive_output(actualizar, {"a": slider_a, "b": slider_b})
display(widgets.VBox([slider_a, slider_b, panel]))
```
